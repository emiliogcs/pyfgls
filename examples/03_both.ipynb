# Serial correlation and heteroskedasticity together

[PG26] Supplement S-4: AR(1) errors whose innovations have variance $x_t^2$. F-C corrects for serial correlation only; F-CH also weights by the Lasso-estimated variances.

```python
import warnings

import numpy as np
import pandas as pd

import pyfgls as pf

warnings.simplefilter("ignore", pf.exceptions.PyFGLSWarning)
```

```python
from pyfgls.simulate import combined_dgp

sim = combined_dgp(300, ar=(0.5,), skedastic='x2', random_state=2)
res = pf.FGLS(sim.y, sim.X, random_state=0).fit()
print(res.summary())
```

```python
res.compare_stages().round(4)
```

An independent check of the analytic standard errors with the sieve bootstrap (few replications here to keep the notebook fast):

```python
boot = pf.FGLS(sim.y, sim.X, random_state=0, cv_folds=5).fit(cov_type='bootstrap', n_boot=49)
pd.DataFrame({'analytic': res.bse, 'bootstrap': boot.bse}).round(4)
```
